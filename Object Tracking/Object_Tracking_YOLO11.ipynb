```python
from ultralytics import YOLO

# Configure the tracking parameters and run the tracker
model = YOLO("yolo11n.pt")
```

## Tracking

```python
!wget -O video.mp4 https://raw.githubusercontent.com/intel-iot-devkit/sample-videos/master/person-bicycle-car-detection.mp4
```

```python
results = model.track(source="video.mp4", conf=0.3, iou=0.5, save=True)
```

```python
results = model.track(source="video.mp4", show=True, tracker="bytetrack.yaml") #with ByteTrack
```
