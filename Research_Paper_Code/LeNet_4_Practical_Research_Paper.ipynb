## LeNet 5 with MNIST Keras

### Imports

```python
from tensorflow import keras
from keras.datasets import mnist
from tensorflow.keras.layers import Dense, Flatten, Conv2D, AveragePooling2D, MaxPooling2D
from keras.layers import Dense, Flatten
from keras.models import Sequential
```

### Loading the dataset and performing splitting

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
# performing reshaping operation
x_train = x_train.reshape(x_train.shape[0], 28, 28, 1)
x_test = x_test.reshape(x_test.shape[0], 28, 28, 1)
```

```python
x_train.shape, x_test.shape
```

### Normalization

```python
x_train = x_train / 255
x_test = x_test / 255
```

### One hot encoding the labels

```python
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)
```

```python
y_test[100]
```

### Network Structure

```python
# Building the Model Architecture

model = Sequential()

model.add(Conv2D(6, kernel_size=(5, 5), strides=(1, 1), activation='relu', input_shape=(28, 28, 1)))  # 24*24*
model.add(MaxPooling2D(pool_size=(2, 2))) # 12*12*6

model.add(Conv2D(16, kernel_size=(5, 5), activation='relu'))  # 8*8*16
model.add(MaxPooling2D(pool_size=(2, 2))) # 4*4*16
model.add(Conv2D(120, kernel_size=(5, 5), activation='relu', padding='same')) # 4*4*120

model.add(Flatten())
model.add(Dense(84, activation='relu'))
model.add(Dense(10, activation='softmax'))

```

```python
model.summary()
```

### Model Compilation

```python
model.compile(loss=keras.metrics.categorical_crossentropy, optimizer=keras.optimizers.Adam(), metrics=['accuracy'])
model.fit(x_train, y_train, batch_size=128, epochs=10, verbose=1, validation_data=(x_test, y_test))
```

### Result

```python
score = model.evaluate(x_test, y_test)
print('Test Loss:', score[0])
print('Test accuracy:', score[1])
```
