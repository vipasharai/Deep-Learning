# Creating random tensors
-  Random tensors are tensors of some abitrary size which contain random number.

### why should we want to create random tensors?
- This is what neural network use to intialize their weights (patterns) that they are trying to learn in the data.

Example - the process of numeral network learning often involves taking a random n- dimensional array of number and refining them until they represnt some kind of pattern (a compressed way to represent the original data )

 ## How Network learns ?
 1. Initalize with random weights(Only at beginning)
 2. Show example
 3. Update Representation outputs.
 4. Repeat with more example



```python
# Create two random (but the same ) tensors

random_1=tf.random.Generator.from_seed(42) # set the seed for reproductivity
random_1= random_1.normal(shape=(3,2))  # create tensor from a normal distribution

random_2=tf.random.Generator.from_seed(42)
random_2=random_2.normal(shape=(3,2))

# Are they equal?
random_1, random_2, random_1 == random_2
```

The random tensor we have made are actually pseudorandom numbers

```python
# Create two random (and different) tensors

random_3=tf.random.Generator.from_seed(42)
random_3= random_3.normal(shape=(3,2))

random_4=tf.random.Generator.from_seed(11)
random_4=random_4.normal(shape=(3,2))

# Are they equal?
random_3, random_4, random_1 == random_3, random_3 == random_4
```

```python
# Shuffle a tensor (valuable for when we want to shuffle your data)

not_shuffled = tf.constant([[10,7],
                            [3,4],
                            [2,5]])
# Gets different results each time
tf.random.shuffle(not_shuffled)
```

```python
# shuffle in the same order every time using the seed parameter (won't actually be the same)

tf.random.shuffle(not_shuffled, seed=42)
```

```python
# shuffle in the same order every time

# set the global random seed
tf.random.set_seed(42)

#  set the operation random seed
tf.random.shuffle(not_shuffled, seed=42)
```

```python
# set the global random seed
tf.random.set_seed(42)  # if you comment this out you'll get different results

# set the operation random seed
tf.random.shuffle(not_shuffled)
```

## Matrix Multiplication
One of the most common operations in ml algorithm is matrix multiplication.
TensorFlow implemenrs this matrix multiplication functionally in the tf.matmul()

The two main rules for matrix

1.   The inner dimension mush must:
-  - (3,5) @ (3,5) won't work
- - (5,3) @ (3,5) will work
2.   The resulting matrix has the shape of the outier dimesnsion:
- - (5,3) @ (3,5) -> (5,5)
- - (3,5) @ (5,3) -> (3,3)

- Note : '@' in python is the symbol for matrix multiplication




```python
# Matrix multipication in Tensorflow
print(tensor)
tf.matmul(tensor, tensor)
```

```python
# Matrix multiplication with python operator '@'
tensor @ tensor
```

```python
# Find the minimum elent position of f
tf.argmin(f)
```

```python
# find the maximum element position of f
print(f"The maximum value of f is at position:{tf.argmax(f).numpy()}")
print(f"The maximum value of f is :{tf.reduce_max(f).numpy()}")
print(f"Using tf.argmax() to index f, he maximum value of f is :{f[tf.argmax(f)].numpy()}")
print(f"Are the two max values the same (they should be)? :{f[tf.argmax(f)].numpy() == tf.reduce_max(f).numpy()}")
```

##Using @tf.function

```python
# cretae a simple function
def  function(x,y):
  return x**2+y
x=tf.constant(np.arange(0,10))
y=tf.constant(np.arange(10,20))
function(x,y)
```

```python
# create the same function and decorate it with tf.function
@tf.function
def tf_function(x,y):
  return x**2+y
tf_function(x,y)
```
